```python
url = "https://serpapi.com/search.json"

all_api_data = []

for keyword in keywords:

    params = {
        "engine": "google_shopping",
        "q": keyword,
        "location": "India",
        "api_key": api
    }

    response = requests.get(
        url,
        params=params
    )

    data = response.json()

    results = data.get(
        "shopping_results",
        []
    )

    for item in results:

        record = {
            "Keyword": keyword,
            "title": item.get("title"),
            "price": item.get("extracted_price"),
            "raw_price": item.get("extracted_original_price"),
            "rating": item.get("rating"),
            "Reviews": item.get("reviews"),
            "platform": item.get("source"),
            "position": item.get("position"),
            "Delivery": item.get("delivery"),
            "Link": item.get("product_link"),
            "thumbnail": item.get("thumbnail")
        }

        all_api_data.append(record)
```

EDA Questions

🟦 General Market Analysis

Ques.1. How many total products are available per keyword?

```python
query = """
SELECT 
    keyword,
    COUNT(*) AS total_products
FROM products
GROUP BY keyword
ORDER BY total_products DESC;
"""
result = pd.read_sql_query(query, conn)

print(result)
```

Ques.2. What is the overall average price of products?

```python
query = """
SELECT 
    ROUND(AVG(price), 2) AS average_price
FROM products
WHERE price IS NOT NULL;
"""
result = pd.read_sql_query(query, conn)
print(result)
```

Ques.3. What is the distribution of price across all products?

```python
query = """
SELECT
    CASE
        WHEN price < 1000 THEN 'Below 1000'
        WHEN price < 5000 THEN '1000 - 4999'
        WHEN price < 10000 THEN '5000 - 9999'
        WHEN price < 25000 THEN '10000 - 24999'
        WHEN price < 50000 THEN '25000 - 49999'
        WHEN price < 100000 THEN '50000 - 99999'
        ELSE '100000+'
    END AS price_bucket,
    COUNT(*) AS product_count
FROM products
WHERE price IS NOT NULL
  AND price > 0
GROUP BY price_bucket
ORDER BY
    CASE price_bucket
        WHEN 'Below 1000' THEN 1
        WHEN '1000 - 4999' THEN 2
        WHEN '5000 - 9999' THEN 3
        WHEN '10000 - 24999' THEN 4
        WHEN '25000 - 49999' THEN 5
        WHEN '50000 - 99999' THEN 6
        WHEN '100000+' THEN 7
    END;
"""
result = pd.read_sql_query(query, conn)
print(result)
```

Ques.4. What is the average rating across all products?

```python
query = """
SELECT 
    ROUND(AVG(rating), 2) AS average_rating
FROM products
WHERE rating IS NOT NULL;
"""
result = pd.read_sql_query(query, conn)
print(result)
```

Ques.5. How are reviews distributed (low vs high engagement products)?

```python
query = """
SELECT
    CASE
        WHEN Reviews < 100 THEN 'Low Engagement'
        WHEN Reviews < 1000 THEN 'Medium Engagement'
        ELSE 'High Engagement'
    END AS engagement_level,
    COUNT(*) AS product_count
FROM products
WHERE Reviews IS NOT NULL
    AND Reviews >= 0
GROUP BY engagement_level
ORDER BY
    CASE engagement_level
        WHEN 'Low Engagement' THEN 1
        WHEN 'Medium Engagement' THEN 2
        WHEN 'High Engagement' THEN 3
    END;
"""
result = pd.read_sql_query(query, conn)
print(result)
```

🟨 Pricing Analysis

Ques.11. What is the price distribution across different brands?

```python
query = """
SELECT
    brand,
    ROUND(AVG(price), 2) AS average_price,
    ROUND(MIN(price), 2) AS min_price,
    ROUND(MAX(price), 2) AS max_price,
    COUNT(price) AS priced_products
FROM products
WHERE brand IS NOT NULL
    AND price IS NOT NULL
    AND price > 0
GROUP BY brand
ORDER BY average_price DESC;
"""
result = pd.read_sql_query(query, conn)
print(result)
```

Ques.12. How are products distributed across price_range categories?

```python
query = """
SELECT
    price_range,
    COUNT(*) AS product_count
FROM products
WHERE price_range IS NOT NULL
GROUP BY price_range
ORDER BY
    CASE price_range
        WHEN 'Budget' THEN 1
        WHEN 'Mid Range' THEN 2
        WHEN 'Premium' THEN 3
        WHEN 'Luxury' THEN 4
    END;
"""
result = pd.read_sql_query(query, conn)
print(result)
```

Ques.13. Do higher-priced products have better rankings (position)?

```python
query = """
SELECT
    CASE
        WHEN price < 1000 THEN 'Below 1000'
        WHEN price < 5000 THEN '1000 - 4999'
        WHEN price < 10000 THEN '5000 - 9999'
        WHEN price < 25000 THEN '10000 - 24999'
        WHEN price < 50000 THEN '25000 - 49999'
        WHEN price < 100000 THEN '50000 - 99999'
        ELSE '100000+'
    END AS price_bucket,
    COUNT(*) AS ranked_products,
    ROUND(AVG(price), 2) AS avg_price,
    ROUND(AVG(position), 2) AS avg_position
FROM products
WHERE price IS NOT NULL
  AND price > 0
  AND position IS NOT NULL
GROUP BY price_bucket
ORDER BY avg_price;
"""

result = pd.read_sql_query(query, conn)
print(result)
```

🟥 Discount & Offer Analysis

Ques.16. What percentage of products have a non-zero discount?

```python
query = """
SELECT
    ROUND(
        100 * SUM(
            CASE
            WHEN raw_price IS NOT NULL
                AND raw_price > price
            THEN 1
            ELSE 0
        END
        ) / COUNT(*), 2)
    AS discounted_percentage
FROM products
WHERE price IS NOT NULL
    AND price > 0;
"""
result = pd.read_sql_query(query, conn)
print(result)
```

Ques.17. Do discounted products have better rankings (position)?

```python
query = """
SELECT
    CASE
        WHEN raw_price > price THEN 'Discounted'
        ELSE 'Not Discounted'
    END AS discount_status,
    COUNT(*) AS product_count,
    ROUND(AVG(position), 2) AS avg_position
FROM products
WHERE price IS NOT NULL
    AND price > 0
    AND position IS NOT NULL
    AND raw_price IS NOT NULL
GROUP BY discount_status
ORDER BY avg_position ASC;
"""
result = pd.read_sql_query(query, conn)
print(result)
```

Ques.29. Do products with higher reviews rank better?

```python
query = """
SELECT 
    CASE
        WHEN reviews < 100 THEN 'Low Reviews'
        WHEN reviews < 1000 THEN 'Medium Reviews'
        ELSE 'High Reviews'
    END AS review_category,
    COUNT(*) AS product_count,
    ROUND(AVG(reviews), 2) AS average_review,
    ROUND(AVG(position), 2) AS average_position
FROM products
WHERE reviews IS NOT NULL
    AND reviews >= 0
    AND position IS NOT NULL
GROUP BY review_category
ORDER BY average_position ASC;
"""
result = pd.read_sql_query(query, conn)
print(result)
```

Ques.30. Which combination of factors (price, rating, reviews) influences top-ranking products?

```python
query = """
SELECT
    CASE
        WHEN position <= 10 THEN 'Top 10'
        ELSE 'Beyond Top 10'
    END AS ranking_group,
    COUNT(*) AS product_count,
    ROUND(AVG(price), 2) AS average_price,
    ROUND(AVG(rating), 2) AS average_rating,
    ROUND(AVG(reviews), 2) AS average_reviews,
    ROUND(AVG(visibility_score), 2) AS average_visibility_score
FROM products
WHERE position IS NOT NULL
    AND price IS NOT NULL
    AND price > 0
    AND rating IS NOT NULL
    AND rating BETWEEN 1 AND 5
    AND reviews IS NOT NULL
    AND reviews >= 0
GROUP BY ranking_group
ORDER BY
    CASE ranking_group
        WHEN 'Top 10' THEN 1
        ELSE 2
    END;
"""
result = pd.read_sql_query(query, conn)
print(result)
```
