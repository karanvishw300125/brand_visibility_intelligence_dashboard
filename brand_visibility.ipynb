```python
from dotenv import load_dotenv
import os

load_dotenv()

api = os.getenv("api")

print(api)
```

```python
keywords = [
    "laptop",
    "television",
    "phone",
    "washing machine",
    "microwave",
    "fridge",
    "tablet",
    "chair",
    "camera",
    "printer"
]
```

```python
url = "https://serpapi.com/search.json"

api_data = []

for keyword in keywords:
    print(f"Searching: {keyword}")
    keyword_count = 0
    start = 0

    while keyword_count < 100:
        params = {
            "engine": "google_shopping",
            "q": keyword,
            "location": "India",
            "api_key": api,
            "start": start,
            "num": 40
        }

        response = requests.get(url,params=params)
        data = response.json()

        results = data.get(
            "shopping_results",
            []
        )

        if not results:
            print(f"No more results at start={start}")
            break

        for item in results:

            record = {
                "Keyword": keyword,
                "title": item.get("title"),
                "price": item.get("extracted_price"),
                "raw_price": item.get("price"),
                "rating": item.get("rating"),
                "reviews": item.get("reviews"),
                "platform": item.get("source"),
                "position": item.get("position"),
                "delivery": item.get("delivery"),
                "link": item.get("link"),
                "thumbnail": item.get("thumbnail")
            }

            api_data.append(record)

        keyword_count += len(results)
        start += len(results)
        print(f"Page fetched | start = {start - len(results)} | Got {len(results)} results | Total: {keyword_count}")

        time.sleep(1)
 
    print(f"{keyword}: {keyword_count} results collected")

api_df = pd.DataFrame(api_data)
print(f"\n{'='*50}")
print(f"Total results collected: {len(api_df)}")
print(api_df.groupby("Keyword").size())
```

```python
def parse_numeric(series):
    """Convert numbers such as '₹12,999', '$1,299.00', '1.2K' to numeric.
    Ambiguous values such as 'many' and 'Not Available' become NaN.
    """
    def one(value):
        if pd.isna(value):
            return np.nan
        if isinstance(value, (int, float, np.integer, np.floating)):
            return float(value) if np.isfinite(value) else np.nan
        s = str(value).strip().lower().replace(",", "")
        if s in {"", "na", "n/a", "nan", "none", "null", "many", "not available", "unknown", "-"}:
            return np.nan
        multiplier = 1.0
        if re.search(r"\d\s*k\b", s):
            multiplier = 1_000.0
            s = re.sub(r"(?<=\d)\s*k\b", "", s)
        elif re.search(r"\d\s*m\b", s):
            multiplier = 1_000_000.0
            s = re.sub(r"(?<=\d)\s*m\b", "", s)
        s = re.sub(r"[^0-9.\-]", "", s)
        if s in {"", "-", ".", "-.", "--"}:
            return np.nan
        try:
            return float(s) * multiplier
        except ValueError:
            return np.nan

    return series.map(one).astype("float64")


def clean_title(series):
    s = series.astype("string")

    # Remove repeated exclamation and hash symbols
    s = s.str.replace(r"!+", "", regex=True)
    s = s.str.replace(r"#+", "", regex=True)

    # Remove emojis using Unicode code points
    s = s.map(
        lambda x: (
            "".join(
                char for char in x
                if not (
                    0x1F300 <= ord(char) <= 0x1FAFF
                )
            )
            if pd.notna(x)
            else pd.NA
        ),
        na_action=None
    ).astype("string")

    # Normalize extra spaces
    s = s.str.replace(r"\s+", " ", regex=True).str.strip()

    # Convert empty titles to missing values
    return s.replace("", pd.NA)


def clean_category(series, default="Unknown"):
    s = series.astype("string").str.strip().str.replace(r"\s+", " ", regex=True)
    s = s.replace("", pd.NA).fillna(default)
    return s


def clean_platform(series):
    s = clean_category(series).str.lower()
    s = s.str.replace(r"\s+", " ", regex=True).str.strip()
    aliases = {
        "amazon.com": "amazon",
        "amazon.in": "amazon",
        "flipkart.com": "flipkart",
        "croma.com": "croma",
        "reliance digital": "reliance digital",
    }
    return s.replace(aliases)
```

```python
# 1. KNOWN BRANDS

KNOWN_BRANDS = [
    "Hewlett Packard", "HP", "Samsung", "Apple", "Microsoft",
    "Lenovo", "Dell", "LG", "Sony", "Asus", "Acer", "MSI",
    "Razer", "Alienware", "Google", "Motorola", "OnePlus",
    "Xiaomi", "Redmi", "Realme", "Oppo", "Vivo", "Nokia",
    "Nothing", "Panasonic", "Toshiba", "Whirlpool", "Bosch",
    "Haier", "Godrej", "IFB", "Voltas", "Daikin", "Carrier",
    "Hitachi", "Blue Star", "Lloyd", "Mitsubishi", "Philips",
    "Bajaj", "Havells", "Usha", "Orient", "Crompton",
    "Prestige", "Pigeon", "Wonderchef", "Dyson", "Eureka Forbes",
    "Kent", "Aquaguard", "Canon", "Nikon", "Fujifilm", "JBL",
    "Bose", "Boat", "boAt", "Jabra", "Sennheiser", "Logitech",
    "TP-Link", "Netgear", "Epson", "Brother", "Puma", "Nike",
    "Adidas", "Reebok", "Ikea", "Wakefit", "Sleepwell",
    "Urban Ladder", "Nilkamal", "Cello", "Whiteduck"
]

# 2. BRAND ALIASES

BRAND_ALIASES = {
    "hewlett packard": "HP",
    "hp": "HP",
    "boat": "boAt",
    "redmi": "Xiaomi"
}

# 3. GENERIC WORDS

GENERIC_START_WORDS = {
    "best", "new", "portable", "wireless", "smart",
    "electric", "digital", "premium", "heavy", "office",
    "ergonomic", "gaming", "convection", "microwave",
    "laptop", "television", "tv", "phone", "washer",
    "fridge", "refrigerator", "table", "chair", "air",
    "power", "men", "women", "home", "kitchen", "large",
    "small", "medium", "automatic", "inverter", "mini",
    "inch", "inches", "modern", "model", "series", "ge",
    "black", "steel", "plastic", "wooden", "pack", "combo",
    "bundle", "unknown", "union", "cuft", "cm", "mm", "k",
    "kg", "gm", "mg", "ml", "gb", "tb", "mah", "watt", "GE",
    "volt", "hz", "tablet", "universal", "replacement", "K"
}

# Match longer brand names first
KNOWN_BRANDS_SORTED = sorted(
    set(KNOWN_BRANDS),
    key=len,
    reverse=True
)

# 4. BRAND EXTRACTION FUNCTION

def extract_brand(title):

    if pd.isna(title):
        return "Unknown"

    title = str(title).strip()

    if not title:
        return "Unknown"

    matches = []

    for brand in KNOWN_BRANDS_SORTED:

        pattern = (
            r"(?<![A-Za-z0-9])"
            + re.escape(brand)
            + r"(?![A-Za-z0-9])"
        )

        match = re.search(
            pattern,
            title,
            flags=re.IGNORECASE
        )

        if match:
            canonical = BRAND_ALIASES.get(
                brand.lower(),
                brand
            )

            matches.append(
                (match.start(), -len(brand), canonical)
            )

    # Prefer the earliest brand mention; longer match wins
    # if two matches start at the same position.
    if matches:
        matches.sort()
        return matches[0][2]

    # Conservative fallback: first token only
    # if it is not a generic product-description word.
    tokens = re.findall(
        r"[A-Za-z][A-Za-z0-9&.+-]*",
        title
    )

    if tokens and tokens[0].lower() not in GENERIC_START_WORDS:
        return tokens[0]

    return "Unknown"


# 5. CHECK REQUIRED COLUMNS

required_columns = [
    "title", "Keyword", "platform"
]

missing_columns = [
    col for col in required_columns
    if col not in clean_df.columns
]

if missing_columns:
    raise KeyError(
        f"Missing required columns: {missing_columns}"
    )

# Keep a backup of the existing brand column
if "brand" in clean_df.columns:
    clean_df["brand_original"] = clean_df["brand"].copy()

# ==========================================
# 6. CLEAN TITLE AND RE-EXTRACT BRAND
# ==========================================

clean_df["title"] = (
    clean_df["title"]
    .astype("string")
    .str.strip()
)

clean_df["brand"] = (
    clean_df["title"]
    .map(extract_brand)
    .astype("string")
)

# ==========================================
# 7. FLAG SUSPICIOUS BRAND VALUES
# ==========================================

SUSPICIOUS_WORDS = {
    "small", "medium", "large", "xl", "xxl", "mini",
    "big", "cu", "inch", "inches", "cm", "mm", "feet",
    "ft", "cuft", "kg", "kgs", "gram", "grams", "gm",
    "mg", "liter", "litre", "liters", "litres", "ml",
    "pack", "packs", "piece", "pieces", "pcs", "att",
    "set", "combo", "bundle", "pair", "count", "gb",
    "tb", "mb", "mah", "watt", "watts", "kw", "volt",
    "volts", "hz", "mp", "mpx", "modern", "series",
    "wireless", "wired", "bluetooth", "rechargeable",
    "portable", "foldable", "ergonomic", "adjustable",
    "digital", "electric", "automatic", "manual",
    "stainless", "steel", "plastic", "wooden", "black",
    "premium", "classic", "standard", "regular", "cafe",
    "model", "heavy", "duty", "new", "latest", "original",
    "tablet", "universal", "compatible", "replacement",
    "unknown", "union", "K", "ge", "GE", "k"
}

brand_normalized = (
    clean_df["brand"]
    .astype("string")
    .str.strip()
    .str.lower()
)

# Flag exact generic matches, excluding Unknown from this flag
suspicious_mask = (
    brand_normalized.isin(SUSPICIOUS_WORDS)
    & brand_normalized.ne("unknown")
)

# Convert suspicious fallback values to Unknown
clean_df.loc[suspicious_mask, "brand"] = "Unknown"

# ==========================================
# 8. DISPLAY SUSPICIOUS / UNKNOWN RECORDS
# ==========================================

unknown_mask = (
    clean_df["brand"]
    .astype("string")
    .str.strip()
    .str.lower()
    .eq("unknown")
)

print("=" * 60)
print("BRAND CLEANING SUMMARY")
print("=" * 60)

print("Total rows:", len(clean_df))
print("Distinct identified brands:",
      clean_df.loc[~unknown_mask, "brand"].nunique())
print("Unknown brand rows:", int(unknown_mask.sum()))

print("\nBrand frequency:")
display(
    clean_df["brand"]
    .value_counts()
    .rename_axis("brand")
    .reset_index(name="count")
    .head(50)
)

print("\nUnknown brand records for manual inspection:")
display(
    clean_df.loc[
        unknown_mask,
        ["Keyword", "title", "brand", "platform"]
    ].head(100)
)

print("\nSuspicious values detected before replacement:")
display(
    clean_df.loc[
        suspicious_mask,
        ["Keyword", "title", "brand_original", "platform"]
    ].head(100)
)

# ==========================================
# 9. CHECK DUPLICATES WITHOUT DELETING
# ==========================================

duplicate_columns = [
    "Keyword", "title", "platform", "price"
]

if all(col in clean_df.columns for col in duplicate_columns):

    duplicate_mask = clean_df.duplicated(
        subset=duplicate_columns,
        keep="first"
    )

    print("\nDuplicate rows detected:",
          int(duplicate_mask.sum()))

    display(
        clean_df.loc[
            duplicate_mask,
            ["Keyword", "title", "brand", "platform", "price"]
        ].head(50)
    )

print("\nBrand cleaning complete. No rows were deleted.")
```

Ques.3. What is the distribution of price across all products?

```python
query = """
SELECT
    CASE
        WHEN price < 1000 THEN 'Below 1000'
        WHEN price < 5000 THEN '1000 - 4999'
        WHEN price < 10000 THEN '5000 - 9999'
        WHEN price < 25000 THEN '10000 - 24999'
        WHEN price < 50000 THEN '25000 - 49999'
        WHEN price < 100000 THEN '50000 - 99999'
        ELSE '100000+'
    END AS price_bucket,
    COUNT(*) AS product_count
FROM products
WHERE price IS NOT NULL
  AND price > 0
GROUP BY price_bucket
ORDER BY
    CASE price_bucket
        WHEN 'Below 1000' THEN 1
        WHEN '1000 - 4999' THEN 2
        WHEN '5000 - 9999' THEN 3
        WHEN '10000 - 24999' THEN 4
        WHEN '25000 - 49999' THEN 5
        WHEN '50000 - 99999' THEN 6
        WHEN '100000+' THEN 7
    END;
"""
result = pd.read_sql_query(query, conn)
print(result)
```

🟥 Discount & Offer Analysis

Ques.16. What percentage of products have a non-zero discount?

```python
query = """
SELECT
    ROUND(
        100 * SUM(
            CASE
            WHEN raw_price IS NOT NULL
                AND raw_price > price
            THEN 1
            ELSE 0
        END
        ) / COUNT(*), 2)
    AS discounted_percentage
FROM products
WHERE price IS NOT NULL
    AND price > 0;
"""
result = pd.read_sql_query(query, conn)
print(result)
```

Ques.17. Do discounted products have better rankings (position)?

Ques.30. Which combination of factors (price, rating, reviews) influences top-ranking products?

```python
query = """
SELECT
    CASE
        WHEN rank <= 10 THEN 'Top 10'
        ELSE 'Beyond Top 10'
    END AS ranking_group,
    COUNT(*) AS product_count,
    ROUND(AVG(price), 2) AS average_price,
    ROUND(AVG(rating), 2) AS average_rating,
    ROUND(AVG(reviews), 2) AS average_reviews,
    ROUND(AVG(visibility_score), 2) AS average_visibility_score
FROM products
WHERE rank IS NOT NULL
    AND price IS NOT NULL
    AND price > 0
    AND rating IS NOT NULL
    AND rating BETWEEN 1 AND 5
    AND reviews IS NOT NULL
    AND reviews >= 0
GROUP BY ranking_group
ORDER BY
    CASE ranking_group
        WHEN 'Top 10' THEN 1
        ELSE 2
    END;
"""
result = pd.read_sql_query(query, conn)
print(result)
```
